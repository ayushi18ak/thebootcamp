# Image Editing Using OpenCV

**Aim:** To perform basic image editing and processing operations using OpenCV-Python.

**Operations:** Resizing, cropping, grayscale, brightness, blur, edge detection, shapes, and text.


```python
import cv2

print(cv2.__version__)# Image Editing Using OpenCV
```

```python
import cv2

image = cv2.imread("photo.jpeg")

if image is None:
    print("Image not found")
else:
    print("Image loaded successfully!")
    print("Image size:", image.shape)
```

```python
import cv2

image = cv2.imread("photo.jpeg")

resized = cv2.resize(image, (600, 400))

cv2.imwrite("resized_image.jpeg", resized)

print("Image resized successfully!")
```

```python
from matplotlib import pyplot as plt

plt.imshow(cv2.cvtColor(resized, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
cropped = image[50:350, 100:500]

cv2.imwrite("cropped_image.jpeg", cropped)

print("Image cropped successfully!")
```

```python
plt.imshow(cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

cv2.imwrite("grayscale_image.jpeg", gray)

print("Image converted to grayscale successfully!")
```

```python
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
bright = cv2.convertScaleAbs(image, alpha=1.2, beta=40)

cv2.imwrite("bright_image.jpeg", bright)

print("Brightness and contrast adjusted successfully!")
```

```python
plt.imshow(cv2.cvtColor(bright, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
blurred = cv2.GaussianBlur(image, (7, 7), 0)

cv2.imwrite("blurred_image.jpeg", blurred)

print("Blur applied successfully!")
```

```python
plt.imshow(cv2.cvtColor(blurred, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(gray, 100, 200)

cv2.imwrite("edges_image.jpeg", edges)

print("Edge detection completed successfully!")
```

```python
plt.imshow(edges, cmap="gray")
plt.axis("off")
plt.show()
```

```python
edited = image.copy()

# Draw a rectangle
cv2.rectangle(edited, (50, 50), (400, 300), (0, 255, 0), 3)

# Draw a circle
cv2.circle(edited, (550, 200), 80, (255, 0, 0), 3)

# Add text
cv2.putText(
    edited,
    "OpenCV Image Editing",
    (50, 380),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0, 0, 255),
    2
)

# Save the edited image
cv2.imwrite("final_edited_image.jpeg", edited)

print("Shapes and text added successfully!")
```

```python
plt.figure(figsize=(10, 6))
plt.imshow(cv2.cvtColor(edited, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```
