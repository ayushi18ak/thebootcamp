```python
import matplotlib.pyplot as plt
import pandas as pd

# Kumbh Mela Data
data = {
    "Day": ["Day 1", "Day 2", "Day 3", "Day 4", "Day 5", "Day 6", "Day 7"],
    "Visitors": [250000, 320000, 450000, 380000, 520000, 600000, 700000],
    "City": ["Delhi", "Mumbai", "Delhi", "Lucknow", "Mumbai", "Prayagraj", "Lucknow"],
    "Participants": [12000, 15000, 18000, 10000, 14000, 25000, 16000]
}

df = pd.DataFrame(data)

print(df)


# ------------------------------------------------
# 1. Line Plot - Number of Visitors Over Days
# ------------------------------------------------

plt.figure(figsize=(8, 5))

plt.plot(
    df["Day"],
    df["Visitors"],
    marker="o",
    linewidth=2,
    label="Number of Visitors"
)

plt.title("Kumbh Mela Visitors Over Different Days")
plt.xlabel("Day")
plt.ylabel("Number of Visitors")

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.tight_layout()

# Save the plot
plt.savefig("kumbh_mela_visitors.png", dpi=300)

plt.show()


# ------------------------------------------------
# 2. Bar Chart - Participants by City
# ------------------------------------------------

city_data = df.groupby("City")["Participants"].sum()

plt.figure(figsize=(8, 5))

plt.bar(
    city_data.index,
    city_data.values,
    label="Participants"
)

plt.title("Distribution of Kumbh Mela Participants by City")
plt.xlabel("City")
plt.ylabel("Number of Participants")

plt.grid(axis="y", linestyle="--", alpha=0.6)
plt.legend()

plt.tight_layout()

# Save the plot
plt.savefig("kumbh_mela_participants.png", dpi=300)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate values
x = np.linspace(0, 2 * np.pi, 500)
y = np.sin(x)

# Plot sine wave
plt.figure(figsize=(10, 6))

plt.plot(
    x,
    y,
    linewidth=2,
    label="Sine Wave"
)

plt.title("Sine Wave", fontsize=16, fontweight="bold")
plt.xlabel("X")
plt.ylabel("sin(X)")

plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.tight_layout()
plt.show()
```
