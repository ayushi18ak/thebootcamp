# Digital Signature Using OpenCV

**Aim:** To create and save a digital signature using OpenCV and mouse events.

```python
import cv2
import numpy as np

# Create a white canvas
canvas = np.ones((500, 800, 3), dtype=np.uint8) * 255

drawing = False
last_x, last_y = -1, -1

def draw_signature(event, x, y, flags, param):
    global drawing, last_x, last_y

    if event == cv2.EVENT_LBUTTONDOWN:
        drawing = True
        last_x, last_y = x, y

    elif event == cv2.EVENT_MOUSEMOVE:
        if drawing:
            cv2.line(canvas, (last_x, last_y), (x, y), (0, 0, 0), 3)
            last_x, last_y = x, y

    elif event == cv2.EVENT_LBUTTONUP:
        drawing = False
        last_x, last_y = -1, -1


# Create window
cv2.namedWindow("Digital Signature")
cv2.setMouseCallback("Digital Signature", draw_signature)

print("Draw your signature using the mouse.")
print("Press S to save the signature.")
print("Press C to clear the canvas.")
print("Press ESC to exit.")

while True:
    cv2.imshow("Digital Signature", canvas)

    key = cv2.waitKey(1) & 0xFF

    if key == ord("s"):
        cv2.imwrite("my_signature.png", canvas)
        print("Signature saved as my_signature.png")

    elif key == ord("c"):
        canvas[:] = 255
        print("Canvas cleared.")

    elif key == 27:
        break

cv2.destroyAllWindows()
```

## Result

Successfully created and saved a digital signature using OpenCV mouse events.
