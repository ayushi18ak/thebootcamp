```python
import cv2
import matplotlib.pyplot as plt

image_path = "river_pollution.jfif"

image = cv2.imread(image_path)

if image is None:
    print("Image not found. Check the file name and location.")
else:
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    plt.figure(figsize=(10, 6))
    plt.imshow(image_rgb)
    plt.axis("off")
    plt.title("Original River Image")
    plt.show()
```

```python
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model(image, verbose=False)

print("YOLO detection completed!")
print("Total objects detected:", len(results[0].boxes))
```

```python
detected_objects = []

for box in results[0].boxes:
    class_id = int(box.cls[0])
    object_name = model.names[class_id]
    confidence = float(box.conf[0])

    detected_objects.append(object_name)

    print(f"Object: {object_name} | Confidence: {confidence:.2f}")
```

```python
results = model(image, conf=0.10, verbose=False)

print("Detection completed!")
print("Total objects detected:", len(results[0].boxes))
```

```python
detected_objects = []

for box in results[0].boxes:
    class_id = int(box.cls[0])
    object_name = model.names[class_id]
    confidence = float(box.conf[0])

    detected_objects.append(object_name)

    print(f"Object: {object_name} | Confidence: {confidence:.2f}")
```

```python
from collections import Counter

object_counts = Counter(detected_objects)

print("Detected Object Counts:")
print()

for object_name, count in object_counts.items():
    print(f"{object_name}: {count}")
```

```python
annotated_image = results[0].plot()

annotated_image_rgb = cv2.cvtColor(
    annotated_image,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(12, 7))
plt.imshow(annotated_image_rgb)
plt.axis("off")
plt.title("River Image - YOLO Object Detection")
plt.show()
```

```python
waste_classes = ["bottle"]

waste_count = 0

for object_name in detected_objects:
    if object_name in waste_classes:
        waste_count += 1

print("Visible waste objects detected:", waste_count)
```

```python
total_objects = len(detected_objects)

print("=" * 40)
print("     RIVER WASTE MONITORING REPORT")
print("=" * 40)

print(f"Total objects detected : {total_objects}")
print(f"Visible waste detected : {waste_count}")

if waste_count == 0:
    print("Status: No detectable bottle waste found")
else:
    print("Status: Visible waste detected")

print("=" * 40)
```

```python
output_path = "river_waste_detection_result.jpg"

cv2.imwrite(output_path, annotated_image)

print("Result saved as:", output_path)
```
