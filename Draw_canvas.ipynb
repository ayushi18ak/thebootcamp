# Draw Canvas and Distance Analysis

This practical demonstrates basic canvas generation and distance
analysis using OpenCV and NumPy.


```python
import cv2
import numpy as np
import math

print("Libraries imported successfully!")
```

```python
# Create a black canvas
canvas = np.zeros((500, 700, 3), dtype=np.uint8)

# Display the canvas
cv2.imshow("My Canvas", canvas)

# Wait for a key press
cv2.waitKey(0)

# Close the window
cv2.destroyAllWindows()
```

```python
# Create a new black canvas
canvas = np.zeros((500, 700, 3), dtype=np.uint8)

# Define two points
point1 = (150, 200)
point2 = (500, 350)

# Draw the two points
cv2.circle(canvas, point1, 8, (255, 255, 255), -1)
cv2.circle(canvas, point2, 8, (255, 255, 255), -1)

# Display the canvas
cv2.imshow("Two Points on Canvas", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Draw a line between the two points
cv2.line(canvas, point1, point2, (0, 0, 255), 2)

# Display the result
cv2.imshow("Distance Line", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# Calculate distance between the two points
distance = math.sqrt(
    (point2[0] - point1[0])**2 +
    (point2[1] - point1[1])**2
)

print("Distance between Point 1 and Point 2:", distance)
```

```python
# Create the canvas again
canvas = np.zeros((500, 700, 3), dtype=np.uint8)

# Draw the two points
cv2.circle(canvas, point1, 8, (255, 255, 255), -1)
cv2.circle(canvas, point2, 8, (255, 255, 255), -1)

# Draw line between the points
cv2.line(canvas, point1, point2, (0, 0, 255), 2)

# Display distance on the canvas
cv2.putText(
    canvas,
    f"Distance: {distance:.2f} pixels",
    (180, 80),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (255, 255, 255),
    2
)

# Display final result
cv2.imshow("Canvas Distance Analysis", canvas)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Start webcam
cap = cv2.VideoCapture(0)

print("Blue colour detection started.")
print("Show a blue object to the camera.")
print("Press 'q' to quit.")

while True:
    ret, frame = cap.read()

    if not ret:
        print("Could not access camera.")
        break

    # Flip the frame like a mirror
    frame = cv2.flip(frame, 1)

    # Convert BGR image to HSV
    hsv = cv2.cvtColor(frame, cv2.COLOR_BGR2HSV)

    # Define blue colour range
    lower_blue = np.array([100, 150, 0])
    upper_blue = np.array([140, 255, 255])

    # Create mask for blue colour
    mask = cv2.inRange(hsv, lower_blue, upper_blue)

    # Find blue objects
    contours, _ = cv2.findContours(
        mask,
        cv2.RETR_EXTERNAL,
        cv2.CHAIN_APPROX_SIMPLE
    )

    # Check if a blue object is detected
    if contours:
        largest_contour = max(contours, key=cv2.contourArea)

        if cv2.contourArea(largest_contour) > 500:

            x, y, w, h = cv2.boundingRect(largest_contour)

            # Draw rectangle around blue object
            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

            # Display message
            cv2.putText(
                frame,
                "BLUE DETECTED",
                (x, y - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.8,
                (255, 255, 255),
                2
            )

    # Show result
    cv2.imshow("Blue Colour Detection", frame)

    # Press q to quit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()

print("Blue colour detection completed.")
```

## Conclusion

In this practical, OpenCV and NumPy were used to generate a canvas,
mark two points, calculate the distance between them, and perform
blue colour detection using HSV colour segmentation and contours.
