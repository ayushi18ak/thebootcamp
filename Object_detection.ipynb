# Real-Time Object Detection and Counting

**Objective:** To detect objects in real time using YOLO and OpenCV and display the count of each detected object class.

```python
import cv2
from ultralytics import YOLO
from collections import Counter

print("Libraries imported successfully!")
```

```python
model = YOLO("yolov8n.pt")

print("YOLO model loaded successfully!")
```

```python
import numpy as np

# Create a blank test image
test_image = np.zeros((500, 700, 3), dtype=np.uint8)

# Run YOLO detection
results = model(test_image, verbose=False)

print("YOLO detection test completed!")
print("Objects detected:", len(results[0].boxes))
```

```python
cap = cv2.VideoCapture(0)

if cap.isOpened():
    print("Webcam opened successfully!")
else:
    print("Could not open webcam.")
```

```python
success, frame = cap.read()

if success:
    print("Webcam frame captured successfully!")
    print("Frame size:", frame.shape)
else:
    print("Failed to capture frame.")
```

```python
results = model(frame, verbose=False)

print("Object detection completed!")
print("Total objects detected:", len(results[0].boxes))
```

```python
class_names = []

for box in results[0].boxes:
    class_id = int(box.cls[0])
    object_name = model.names[class_id]
    class_names.append(object_name)

print("Detected objects:")
print(class_names)object_counts = Counter(class_names)

print("Class-wise object count:")

for object_name, count in object_counts.items():
    print(object_name, ":", count)
```

```python
# Open webcam again
cap = cv2.VideoCapture(0)

print("Starting Real-Time Object Detection...")
print("Press 'q' to quit.")

while cap.isOpened():

    success, frame = cap.read()

    if not success:
        print("Failed to grab camera frame.")
        break

    # Flip frame horizontally for a mirror-like view
    frame = cv2.flip(frame, 1)

    # Detect objects
    results = model(frame, verbose=False)

    # Store detected object names
    class_names = []

    for box in results[0].boxes:
        class_id = int(box.cls[0])
        object_name = model.names[class_id]
        class_names.append(object_name)

    # Count objects by class
    object_counts = Counter(class_names)

    # Draw YOLO detection boxes
    annotated_frame = results[0].plot()

    # Display class-wise counts
    y_position = 40

    for object_name, count in object_counts.items():

        text = f"{object_name}: {count}"

        cv2.putText(
            annotated_frame,
            text,
            (20, y_position),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (0, 255, 0),
            2
        )

        y_position += 30

    # Display total count
    total_objects = len(class_names)

    cv2.putText(
        annotated_frame,
        f"Total Objects: {total_objects}",
        (20, y_position + 10),
        cv2.FONT_HERSHEY_SIMPLEX,
        0.7,
        (0, 255, 255),
        2
    )

    # Show webcam
    cv2.imshow(
        "Real-Time Object Detection and Counting",
        annotated_frame
    )

    # Press q to quit
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

# Release resources
cap.release()
cv2.destroyAllWindows()

print("Camera feed closed successfully.")
```

```python
# Open webcam
cap = cv2.VideoCapture(0)

print("Starting detection. Press 'q' to quit.")

while cap.isOpened():

    success, frame = cap.read()

    if not success:
        break

    # Detect objects
    results = model(frame, verbose=False)

    # Get detected object names
    class_names = []

    for box in results[0].boxes:
        class_id = int(box.cls[0])
        object_name = model.names[class_id]
        class_names.append(object_name)

    # Print detected objects in Jupyter
    print("Detected:", class_names)

    # Display detections
    annotated_frame = results[0].plot()

    cv2.imshow("YOLO Object Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

cap.release()
cv2.destroyAllWindows()

print("Detection stopped.")
```
