```python
import pandas as pd
import numpy as np

# For reproducible random values
np.random.seed(42)

# Number of records
n = 100

# Create dataset
data = {
    "Dancer_ID": range(1, n + 1),

    "Age": np.random.randint(16, 30, n),

    "Experience_Years": np.round(
        np.random.uniform(0.5, 10, n), 1
    ),

    "Practice_Hours": np.round(
        np.random.uniform(1, 15, n), 1
    ),

    "Practice_Days": np.random.randint(1, 8, n),

    "Warmup_Minutes": np.random.randint(5, 31, n),

    "Sleep_Hours": np.round(
        np.random.uniform(5, 9, n), 1
    ),

    "Energy_Level": np.random.randint(1, 11, n),

    "Practice_Intensity": np.random.choice(
        ["Low", "Medium", "High"],
        n
    ),

    "New_Steps_Learned": np.random.randint(1, 16, n),

    "Mistakes_Per_Session": np.random.randint(0, 21, n),

    "Mood": np.random.choice(
        ["Happy", "Neutral", "Tired"],
        n
    ),

    "Previous_Score": np.round(
        np.random.uniform(40, 95, n), 1
    )
}

# Convert dictionary into DataFrame
df = pd.DataFrame(data)


# --------------------------------------------------
# Create Performance Score
# --------------------------------------------------

df["Performance_Score"] = (
    30
    + df["Practice_Hours"] * 2.5
    + df["Practice_Days"] * 1.5
    + df["Warmup_Minutes"] * 0.3
    + df["Sleep_Hours"] * 2
    + df["Energy_Level"] * 1.5
    + df["Experience_Years"] * 1.5
    + df["New_Steps_Learned"] * 0.5
    + df["Previous_Score"] * 0.2
    - df["Mistakes_Per_Session"] * 1.2
)

# Add some random variation
df["Performance_Score"] += np.random.normal(0, 5, n)

# Keep score between 0 and 100
df["Performance_Score"] = (
    df["Performance_Score"]
    .clip(0, 100)
    .round(1)
)


# --------------------------------------------------
# Create ML Target
# --------------------------------------------------

df["Performance_Level"] = np.where(
    df["Performance_Score"] >= 70,
    "Good",
    "Needs Improvement"
)


# --------------------------------------------------
# Add a few missing values for data-cleaning practice
# --------------------------------------------------

df.loc[np.random.choice(n, 5, replace=False), "Sleep_Hours"] = np.nan

df.loc[np.random.choice(n, 4, replace=False), "Mood"] = np.nan

df.loc[np.random.choice(n, 3, replace=False), "Practice_Hours"] = np.nan


# --------------------------------------------------
# Save as CSV
# --------------------------------------------------

df.to_csv("dancing_performance_dataset.csv", index=False)

print("CSV file created successfully!")
print("File name: dancing_performance_dataset.csv")

# Display first 10 rows
print("\nFirst 10 rows:")
print(df.head(10))
```
