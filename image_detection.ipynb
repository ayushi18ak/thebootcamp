```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# 1. Read the image (Prachi.jpeg)
img = cv2.imread("Prachi.jpeg")

# 2. Print image details
print("Type:", type(img))
print("Shape:", img.shape)

# 3. Display the image (Jupyter-safe method)
# OpenCV reads images in BGR format, so we convert them to RGB
if img is not None:
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    
    plt.figure(figsize=(6, 6))
    plt.imshow(img_rgb)
    plt.axis('off')
    plt.title("Prachi Image")
    plt.show()
else:
    print("Error: Image could not be loaded. Check that the file name is exactly 'Prachi.jpeg'.")
```

```python
img_flip = cv2.flip(img,0)
cv2.imshow("Photos", img_flip)
cv2.waitKey(0)
```

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
img = cv2.imread("Prachi.jpeg")

# Flip the image vertically
img_flip = cv2.flip(img, 0)

# Convert BGR to RGB for displaying with Matplotlib
img_flip_rgb = cv2.cvtColor(img_flip, cv2.COLOR_BGR2RGB)

# Display the flipped image
plt.imshow(img_flip_rgb)
plt.axis("off")
plt.title("Vertically Flipped Image")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Crop the image
img_crop = img[100:300, 200:500]

# Convert BGR to RGB
img_crop_rgb = cv2.cvtColor(img_crop, cv2.COLOR_BGR2RGB)

# Display the cropped image
plt.imshow(img_crop_rgb)
plt.axis("off")
plt.title("Cropped Image")
plt.show()
```

```python
# Import necessary libraries
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# Ensure necessary NLTK data is downloaded
nltk.download('punkt')
nltk.download('stopwords')
```
